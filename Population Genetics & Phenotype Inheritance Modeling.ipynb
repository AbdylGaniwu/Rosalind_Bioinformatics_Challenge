## 3. Python Implementation Strategy

#### a) Method 1: List Comprehension with zip() and sum()

```python
# Population counts for each genotype pairing [AA-AA, AA-Aa, AA-aa, Aa-Aa, Aa-aa, aa-aa]
couples = [19092, 19846, 17704, 17753, 19005, 18105]

# Expected dominant offspring generated per couple type (2 * P_dominant)
multipliers = [2.0, 2.0, 2.0, 1.5, 1.0, 0.0]

# Compute total expected value via element-wise weighted sum
expected_offspring = sum(c * m for c, m in zip(couples, multipliers))

print(f"Total Expected Dominant Offspring: {expected_offspring}")
```

## b) Method 2: Vectorized Dot Product using NumPy (Linear Algebra / Matrix Multiplication Approach)

```python
import numpy as np

# Sample input strings (ROSALIND dataset)
data = "19092 19846 17704 17753 19005 18105"

# Parse counts into a 1D NumPy vector
couples = np.fromstring(data, sep=" ")

# Expected dominant offspring per couple type
multipliers = np.array([2.0, 2.0, 2.0, 1.5, 1.0, 0.0])

# Compute dot product (weighted linear combination)
result = np.dot(couples, multipliers)    # Result = (19092 x 2.0) + (19846 x 2.0) + (17704 x 2.0) + (17753 x 1.5) + (19005 x 1.0) + (18105 x 0.0) = 158918.5

# Dot product calculates the weighted sum directly
result = np.dot(couples, multipliers)

print(f"Total Expected Dominant Offspring: {result}")
```

## c) Method 3: Parameter-Driven Probability Calculation

```python
# Number of offspring per couple
OFFSPRING = 2 

# Mendelian dominant phenotype probabilities P(A_) per single offspring
p_dominant = [
    1.0,
    1.0,
    1.0,
    0.75,
    0.5,
    0.0,
]

# Calculate total expected dominant offspring
counts = [19092, 19846, 17704, 17753, 19005, 18105]

total_expected = sum(
    count * prob * OFFSPRING for count, prob in zip(counts, p_dominant)
)

print(f"Total Expected Dominant Offspring: {total_expected}")
```

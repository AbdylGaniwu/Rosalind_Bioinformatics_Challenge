# Translating RNA into Protein

### The Genetic Code

- In cellular biology, translation is the process by which messenger RNA (mRNA) sequences are decoded to synthesize proteins, the functional workhorses of the cell.
- Codons: Because RNA is made of 4 nucleotide bases (A, U, C, G) and proteins are composed of 20 distinct amino acids, bases are read in non-overlapping groups of three called codons (4^3 = 64 possible combinations).
- Degeneracy & Redundancy: Multiple distinct codons can map to the same amino acid. For example, CUU, CUC, CUA, and CUG all encode Leucine (L).
- Start & Stop Signals:
  - Translation typically begins at the start codon (AUG), which encodes Methionine (M).
  - Translation terminates when the ribosome encounters one of three stop codons (UAA, UAG, or UGA), which do not code for an amino acid and trigger the release of the newly synthesized polypeptide chain.

- Computational Approach.
  - Codon Mapping: Represent the genetic code using a key-value data structure (such as a Python dictionary) mapping each triplet string to its corresponding single-letter amino acid code or a stop signal.
  - Step Slicing: Iterate through the RNA sequence in non-overlapping increments of 3 bases (e.g., indices 0:3, 3:6, 6:9).
  - Sequence Construction: Translate each triplet sequentially and append the amino acid to the output string until a stop codon or the end of the sequence is reached.

```python
def translate_rna(rna_sequence):
    """Translates an RNA sequence into a protein sequence."""
    codon_table = {
        'UUU': 'F', 'CUU': 'L', 'AUU': 'I', 'GUU': 'V',
        'UUC': 'F', 'CUC': 'L', 'AUC': 'I', 'GUC': 'V',
        'UUA': 'L', 'CUA': 'L', 'AUA': 'I', 'GUA': 'V',
        'UUG': 'L', 'CUG': 'L', 'AUG': 'M', 'GUG': 'V',
        'UCU': 'S', 'CCU': 'P', 'ACU': 'T', 'GCU': 'A',
        'UCC': 'S', 'CCC': 'P', 'ACC': 'T', 'GCC': 'A',
        'UCA': 'S', 'CCA': 'P', 'ACA': 'T', 'GCA': 'A',
        'UCG': 'S', 'CCG': 'P', 'ACG': 'T', 'GCG': 'A',
        'UAU': 'Y', 'CAU': 'H', 'AAU': 'N', 'GAU': 'D',
        'UAC': 'Y', 'CAC': 'H', 'AAC': 'N', 'GAC': 'D',
        'UAA': 'Stop', 'CAA': 'Q', 'AAA': 'K', 'GAA': 'E',
        'UAG': 'Stop', 'CAG': 'Q', 'AAG': 'K', 'GAG': 'E',
        'UGU': 'C', 'CGU': 'R', 'AGU': 'S', 'GGU': 'G',
        'UGC': 'C', 'CGC': 'R', 'AGC': 'S', 'GGC': 'G',
        'UGA': 'Stop', 'CGA': 'R', 'AGA': 'R', 'GGA': 'G',
        'UGG': 'W', 'CGG': 'R', 'AGG': 'R', 'GGG': 'G'
    }

    protein = ""

    # Read sequence 3 bases at a time
    for i in range(0, len(rna_sequence), 3):
        codon = rna_sequence[i:i+3]     
        amino_acid = codon_table[codon]

        if amino_acid == "Stop":
            break

        protein += amino_acid

    return protein
```

```python
# Read input sequence from file
with open("translating_rna_data_sample.txt", "r") as file:
    rna_input = file.read().strip()

# Run translation function
protein_result = translate_rna(rna_input)

# Display final protein output
print(protein_result)
```

## Conclusion & Key Takeaways

1. **Successful Translation Execution**: The script converts raw mRNA sequences into amino acid chains by matching each 3-letter codon to the genetic code.
   
2. **Termination Handling**: The program automatically stops reading as soon as it hits a stop codon (UAA, UAG, or UGA), mimicking how ribosomes stop building proteins in real cells.

3. **Data Structure Efficiency**: By storing the genetic code in a dictionary, the script matches every 3-letter codon to its amino acid in constant time. This ensures the code runs smoothly and processes long RNA strands without slowing down.

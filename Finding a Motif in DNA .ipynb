# Motif in DNA 

- In genomic sequence analysis, a motif is a short recurring DNA sequence pattern that often possesses biological significance (such as transcription factor protein binding sites to DNA or protein domain encodings). Finding where a specific motif appears within a larger sequence of DNA is like searching for a specific keyword in a long paragraph.

## Goal

- The goal of this project is to locate all occurrences of a short DNA motif $t$ within a longer DNA sequence $s$. 
- The solution must account for overlapping occurrences and output the locations using standard biological 1-based indexing.
- Give a sample data containing dna sequence, find all starting positions where t occurs as a contiguous substring inside s.

Given:
- **Input**: Two DNA strings,
  - $s$ (main sequence) a long DNA strong (the "haystack") and
  - $t$ (motif pattern) a shorter DNA string ("needle" you are searching for)
- **Output**: Space-separated integers representing all 1-based starting positions where sequence $t$ matches sequence $s$.

## Implementation & Code

```python
# Method 1: Sliding Window Slicing

# Read sequence input from file
with open("finding_motif_in_dna_sequence.txt", "r") as file:
    s = file.read().strip()

# Target motif pattern
t = "CTCCTGA"

positions = []
t_len = len(t)

for i in range(len(s) - t_len + 1):
    if s[i:i + t_len] == t:
        positions.append(i + 1)   # Convert 0-based Python index to 1-based index

# Display main sequence and matched positions
print(s)
print(f"Motif Start Positions (1-based): {' '.join(map(str, positions))}")
```

## Conclusion & Key Takeaways

- **Biological Relevance**: Finding recurring motifs in genomic datasets is fundamental to identifying biological signals like promoter sequences, binding sites, and restriction sites.
- **0-Based vs. 1-Based Indexing**: While Python counts indices starting at 0, standard biological databases and Rosalind demand 1-based positioning. Offsetting Python indices by +1 aligns output with domain standards.
- **Handling Overlapping Patterns**: Algorithms must step incrementally (by offset 1) rather than skipping ahead by motif length (len(t)) to ensure overlapping occurrences are not missed.

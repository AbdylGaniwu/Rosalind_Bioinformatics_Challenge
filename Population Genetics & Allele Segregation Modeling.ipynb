### Experimental Framework

Consider a diploid population consisting of three distinct genotype classes:
- k Homozygous Dominant (AA): Always contributes a dominant allele (A).
- m Heterozygous (Aa): Has a 50% chance of passing A and a 50% chance of passing a.
- n Homozygous Recessive (aa): Always contributes a recessive allele (a).

### Mathematical Logic (Complementary Probability)

Instead of summing all parent combinations that produce dominant offspring, it is computationally simpler to use complementary probability:
                        P(Dominant Phenotype) = 1 - P(Recessive Offspring)

An offspring can display the recessive phenotype (aa) only if both parents pass a recessive allele (a). The only parental pairings capable of producing an aa child are:
- aa x aa:     Both parents homozygous recessive (100% chance of aa offspring).
- Aa x Aa:     Both parents heterozygous (25% chance of aa offspring).
- Aa x aa:     (or aa x Aa): One heterozygous and one homozygous recessive parent (50% chance of aa offspring).

Given total population size T = k + m + n, the total number of ordered parental selections without replacement is T(T - 1).


### 2. Mathematical Framework & Equation Derivation

#### Weighted Recessive Offspring Count

Summing the weighted outcomes across all recessive-producing parental pairs gives:

- Recessive Outcomes = n(n - 1) x 1.0 + m(m - 1) x 0.25 + (2 x m x n x 0.5)
                        aa x aa             Aa x Aa           Aa x aa/Aa


- Dividing by total ordered pairings T(T - 1) gives the probability of a recessive outcome:
P(Recessive) = n(n - 1) + 0.25 x m(m - 1) + m x n
               -----------------------------------
                          T(T - 1)


- Final Closed-Form Equation:
                    _                                   _
                   | n(n - 1) + 0.25 x m(m - 1) + m x n  |
P(Dominant) = 1 -  | ----------------------------------- |
                   |              T(T - 1)               |  
                   |_                                   _|

## Python Implementation Strategies

## The Goal:

Consider a population defined by genotype parameters k = 29, m = 23, and n = 26, 
where k, m, and n represent homozygous dominant (AA), heterozygous (Aa), and homozygous recessive (aa) individuals, respectively. 

- a) Part A: Determine the total population size T and the total number of ordered parent pairs T(T - 1).
- b) Part B: Calculate the total weighted number of outcomes that yield a homozygous recessive (aa) offspring.
- c) Part C: Use complementary probability to find P({Dominant Phenotype}). Express your final answer rounded to 5 decimal places.

Goal: Given a population containing a mix of homozygous dominant (k), heterozygous (m), and homozygous recessive (n) individuals, calculate the probability that two randomly chosen individuals will produce an offspring that expresses the dominant phenotype.

### Method 1: Pure Python Parameter-Driven Function

```python
def get_dominant_probability(k: int, m: int, n: int) -> float:
    # Total population size
    total_pop = k + m + n
    total_pairs = total_pop * (total_pop - 1)     # Total possible ordered pairings

    # Weighted counts of pairings producing recessive (aa) offspring
    recessive_recessive = n * (n - 1) * 1.0                          # aa x aa -> 100% chance of aa child
    hetero_hetero = m * (m - 1) * 0.25                               # Aa x Aa -> 25% chance of aa child
    hetero_recessive = (
        2 * (m * n) * 0.5
    )  # Aa x aa (or aa x Aa) -> 50% chance of aa child

    # Sum of recessive outcomes
    total_recessive_outcomes = (
        recessive_recessive + hetero_hetero + hetero_recessive
    )

    # Complementary probability: P(Dominant) = 1 - P(Recessive)
    p_dominant = 1 - (total_recessive_outcomes / total_pairs)
    return p_dominant


# Benchmark Dataset (Data from ROSALIND)
k_count, m_count, n_count = 29, 23, 26
result = get_dominant_probability(k_count, m_count, n_count)

print(f"{'Population Size (T):':<25} {k_count + m_count + n_count}")
print("-" * 33)
print(f"{'P(Dominant Phenotype):':<25} {result:.5f}")
```

### Method 2: Dictionary-Mapped Parsing from Dataset File

```python
# Read dataset file
with open("sample_data_probability.txt", "r") as file:
    k, m, n = map(int, file.read().split())

# Define genotype counts
genotypes = {
    "Homozygous Dominant (AA)": k,
    "Heterozygous (Aa)": m,
    "Homozygous Recessive (aa)": n,
}


total_pop = sum(genotypes.values())
total_pairs = total_pop * (total_pop - 1)


# Display genotype breakdown
for label, count in genotypes.items():
    print(f"{label:<26} : {count}")
print("-" * 31)
print(f"{'Total Population:':<26} : {total_pop}")


# Calculate probability of dominant phenotype
recessive_prob = (
    (n * (n - 1) * 1.0) + (m * (m - 1) * 0.25) + (2 * m * n * 0.5)
) / total_pairs
p_dominant = 1 - recessive_prob


print("-" * 31)
print(f"{'P(Dominant Phenotype):':<26} : {p_dominant:.5f}")
```

```python
# Open and read the DNA sequence from the file
with open("rosalind_revc.txt", "r") as file:
    dna_string = file.read().strip()
print(dna_string)    
```

```python
# Check for spaces
has_spaces = " " in dna_string

# Check for any lowercase letters
has_lowercase = any(char.islower()
                    for char in dna_string)

print(f"Contains spaces? {has_spaces}")
print(f"Contains lowercase? {has_lowercase}")
```

```python
# Create a duplicate copy
dna_copy = dna_string
```

```python
def reverse_complement(dna_copy):
    translation_table = str.maketrans("ATCG", "TAGC")

    return dna_copy[::-1].translate(translation_table)
print(reverse_complement(dna_copy))
    
```

### Another Method (The Dictionary Lookup)

```python
# Open and read the DNA sequence from the file
with open("rosalind_revc1.txt", "r") as file:
    dna_string1 = file.read().strip()
print(dna_string1)    
```

```python
# Check for spaces
has_spaces = " " in dna_string1

# Check for any lowercase letters
has_lowercase = any(char.islower()
                    for char in dna_string1)

print(f"Contains spaces? {has_spaces}")
print(f"Contains lowercase? {has_lowercase}")
```

```python
def reverse_complement(dna_copy):
    # Define the base pairings in a dictionary
    pairs = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    
    # Reverse the string, look up each base in the dictionary, and join them
    return "".join(pairs[base] for base in reversed(dna_copy))

reverse_dna = dna_string
print(reverse_complement(reverse_dna))
```

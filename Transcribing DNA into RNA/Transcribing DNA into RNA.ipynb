```python
# Open and read DNA sequence from the file
with open("rosalind_rna.txt", "r") as file:
     dna_string = file.read().strip()
print(dna_string)
```

```python
# Check for space
has_space = " " in dna_string

# Check for lowercase letters
has_lowercase = any(char.islower()
                    for char in dna_string)

print(f"Contains Space? {has_space}")
print(f"Contains Lowercase? {has_lowercase}")
```

```python
# Create a duplicate copy
dna_sequence = dna_string
```

```python
def transcribe_dna_to_rna(dna_string):
    return dna_string.replace("T", "U")

rna_sequence = transcribe_dna_to_rna(dna_sequence)
print(rna_sequence)
```

```python
cross_check_T = rna_sequence.find("T")
print(cross_check_T)  # -1 confirms that there are absolutely no "T" (Thymine)
```

```python
# Very Simple Method
dna_string1 = dna_sequence
print(dna_string1.replace("T", "U"))
```
